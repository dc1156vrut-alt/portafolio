```python
import pandas as pd
import numpy as np

diccionario = {"Nombre":['Jose','Maria','Carlos','Andres','Ramiro','Laura','Mauricio','Enesto'],
               "Cargo": ['Ingeniero','Auxiliar contable','Ingeniero','Auxiliar contable','Ingeniero',
                         'Auxiliar contable','Ingeniero','Auxiliar contable'],
               "Salario": [500,np.nan,200,800,np.nan,500,np.nan,800],
               "Ciudad":['Cali','Bogota','Medellin','Bogota','Cali','Medellin','Cali','Bogota']
              }

df = pd.DataFrame(diccionario)

categorias= df.groupby('Cargo')['Salario'].mean()

# Loop through the DataFrame and update salaries based on the condition
for i in range(len(df)):
  # Check if the salary is not NaN, the salary is greater than the mean for their cargo, and the cargo is 'Auxiliar contable'
  if pd.notna(df.loc[i, 'Salario']) and (df.loc[i,'Salario'] > categorias[df.loc[i,'Cargo']]) and (df.loc[i,'Cargo'] == 'Auxiliar contable'):
      # Replace the salary with the mean salary for 'Auxiliar contable' if the condition is met
      df.loc[i,'Salario'] = categorias['Auxiliar contable']
  # If the salary is NaN and the cargo is 'Auxiliar contable', fill the NaN with the mean salary for 'Auxiliar contable'
  elif pd.isna(df.loc[i, 'Salario']) and (df.loc[i,'Cargo'] == 'Auxiliar contable'):
      df.loc[i, 'Salario'] = categorias['Auxiliar contable']


# Display the updated DataFrame
display(df)
```
