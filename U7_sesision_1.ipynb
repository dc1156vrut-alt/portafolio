Ejercicio 5

```python
lado1= float(input('Ingrese la longitud del lado 1 -> '))
lado2= float(input('Ingrese la longitud del lado 2 -> '))
lado3= float(input('Ingrese la longitud del lado 3 -> '))

if lado1 == lado2 and lado2 == lado3:
  print('El triangulo es equilatero')
elif lado1 == lado2 or lado1 == lado3 or lado2 == lado3:
  print('El triangulo es isosceles')
else:
  print('El triangulo es escaleno')
```

Ejercicio 6

```python
numero_empleados = int(input('ingrese el numero de empleados'))
contador100_300 = 0
contador300 = 0
sueldototal = 0

for i in range(numero_empleados):
 sueldo = float(input('\nIngrese el sueldo del empleado\n'))
 if sueldo >= 100 and sueldo < 300:
  contador100_300 =+ contador100_300 + 1
  sueldototal =+ sueldototal + sueldo
 elif sueldo >= 300 and sueldo <= 500:
  contador300 =+ contador300 + 1
  sueldototal =+ sueldototal + sueldo
 else :

  print('El sueldo esta fuera del rango porfavor ingrese nuevamente un valor entre 100 y 500')

print(f'El numero de empleados con sueldos entre 100 y 300 es -> {contador100_300}')
print(f'El numero de empleados con sueldos entre 300 y 500 es -> {contador300}')
print(f'El gasto total de sueldos de la empresa es -> {sueldototal}')



```

```python
numero_piezas = int(input('Ingrese el numero de piezas a verificar'))
aptas = 0
noaptas = 0
for i in range(numero_piezas):
 longitud = float(input(f'Ingrese la longitud de la pieza {i+1}'))
 if longitud >= 1.20 and longitud <= 1.30:
  aptas += 1
 else:
  noaptas += 1

print(f'El numero de piezas aptas es -> {aptas}')
print(f'El numero de piezas no aptas es -> {noaptas}')
```
